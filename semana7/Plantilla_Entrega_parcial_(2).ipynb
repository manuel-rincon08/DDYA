```python
from collections import deque
import networkx as nx
import matplotlib.pyplot as plt
class Nodo:

    def __init__(self, nombre):
        self.nombre = nombre
        self.vecinos = []

    def conectar(self, nodo_nuevo):
        if nodo_nuevo not in self.vecinos:
            self.vecinos.append(nodo_nuevo)
            nodo_nuevo.vecinos.append(self)
    def nom_vecino(self):
        return sorted([vecino.nombre for vecino in self.vecinos])

def busq_prof(grafo, nom_inicio):
    if nom_inicio not in grafo:
        print(f"Error: El centro '{nom_inicio}' no existe en la red.")
        return
    visit = set()
    recorr = []

    def dfs(nodo_actual):
        if nodo_actual.nombre not in visit:
            visit.add(nodo_actual.nombre)
            recorr.append(nodo_actual.nombre)
            for vecino in nodo_actual.vecinos:
                dfs(vecino)
    dfs(grafo[nom_inicio])
    print(f"Recorrido DFS desde {nom_inicio}: {' -> '.join(recorr)}")

def busq_anch(grafo, nom_inicio, nombre_objetivo=None):
    if nom_inicio not in grafo:
        print(f"Error: El centro '{nom_inicio}' no existe en la red.")
        return
    nodo_inicio = grafo[nom_inicio]
    visit = set([nodo_inicio.nombre])
    cola = deque([(nodo_inicio, [nodo_inicio.nombre])])
    recorr = []
    niv = {0: [nodo_inicio.nombre]}
    dist = {nodo_inicio.nombre: 0}
    ruta = None
    while cola:
        nodo_actual, ruta = cola.popleft()
        recorr.append(nodo_actual.nombre)
        if nombre_objetivo and nodo_actual.nombre == nombre_objetivo:
            ruta = ruta
        for vecino in nodo_actual.vecinos:
            if vecino.nombre not in visit:
                visit.add(vecino.nombre)
                nueva_ruta = ruta + [vecino.nombre]
                cola.append((vecino, nueva_ruta))
                distancia = dist[nodo_actual.nombre] + 1
                dist[vecino.nombre] = distancia
                if distancia not in niv:
                    niv[distancia] = []
                niv[distancia].append(vecino.nombre)
    if nombre_objetivo:
        if ruta:
            conexiones = len(ruta) - 1
            print(f"\nCentro encontrado. Ruta: {' -> '.join(ruta)}")
            print(f"Cantidad de conexiones: {conexiones}")
        else:
            print(f"\nNo existe una ruta hasta el centro solicitado '{nombre_objetivo}'.")
    else:
        print(f"Recorrido BFS desde {nom_inicio}: {' -> '.join(recorr)}")
        print("Niveles de la red:")
        for nivel, nodos_nivel in niv.items():
            print(f"  Nivel {nivel}: {', '.join(nodos_nivel)}")

def impr_grafo_visual(grafo):
   print("\nGenerando visualización gráfica en una nueva ventana...")
   G = nx.Graph()
   for nombre_nodo, objeto_nodo in grafo.items():
       G.add_node(nombre_nodo)
       for vecino in objeto_nodo.vecinos:
           G.add_edge(nombre_nodo, vecino.nombre)
   plt.figure(figsize=(8, 6))
   posiciones = nx.spring_layout(G, seed=42)
   nx.draw(G,posiciones,with_labels=True,node_color='skyblue',node_size=2000,font_size=14,font_weight='bold',font_color='black',edge_color='gray',width=2)
   plt.title("Mapa de la Red de Distribución", fontsize=16, fontweight='bold')
   plt.axis('off')
   plt.show()

def impr_grafo(grafo):
    print("\nRepresentación de la Red (Listas Adyacentes):")
    for nombre in sorted(grafo.keys()):
        nodo = grafo[nombre]
        vecinos_str = ", ".join(nodo.nom_vecino())
        print(f"  [{nodo.nombre}] se conecta con -> {vecinos_str}")
    print("-" * 50)

def agreg_centro(grafo):
    print("\n--- Agregar Nuevo Centro ---")
    nuevo_nombre = input("Ingrese el nuevo centro: ").strip().upper()
    if nuevo_nombre in grafo:
        print("Error: El centro ya existe en la red.")
        return grafo
    nuevo_nodo = Nodo(nuevo_nombre)
    grafo[nuevo_nombre] = nuevo_nodo
    print(f"Centro '{nuevo_nombre}' creado. Ahora ingrese con quién se conecta.")
    print("(Presione Enter sin escribir nada para terminar de agregar conexiones)")
    while True:
        nombre_vecino = input(f"Conectar {nuevo_nombre} con: ").strip().upper()
        if not nombre_vecino:
            break
        if nombre_vecino in grafo:
            nodo_existente = grafo[nombre_vecino]
            nuevo_nodo.conectar(nodo_existente)
            print(f"  [+] Conexión {nuevo_nombre} - {nombre_vecino} agregada correctamente.")
        else:
            print(f"  [X] El centro '{nombre_vecino}' no existe en la red actual.")
    return grafo

def main():
    print("BIENVENIDO")
    grafo = {}
    nombres_iniciales = ["A", "B", "C", "D", "E", "F", "G"]
    for nombre in nombres_iniciales:
        grafo[nombre] = Nodo(nombre)
    grafo["A"].conectar(grafo["B"])
    grafo["A"].conectar(grafo["C"])
    grafo["B"].conectar(grafo["D"])
    grafo["B"].conectar(grafo["E"])
    grafo["C"].conectar(grafo["F"])
    grafo["E"].conectar(grafo["G"])
    grafo["F"].conectar(grafo["G"])
    impr_grafo_visual(grafo)
    impr_grafo(grafo)
    print("\n--- Ejecución Inicial Obligatoria desde 'A' ---")
    busq_anch(grafo, "A")
    print()
    busq_prof(grafo, "A")
    while True:
        print("\nMENU")
        print("1. Añadir nuevo centro a partir de un inicio y recorrer")
        print("2. Buscar un centro a partir de un inicio dado (BFS)")
        print("3. Salir")
        opcion = input("Seleccione una opción (1-3): ").strip()
        if opcion == "1":
            grafo = agreg_centro(grafo)
            impr_grafo(grafo)
            impr_grafo_visual(grafo)
            inicio = input("Ingrese el centro inicial para realizar los recorridos: ").strip().upper()
            if inicio in grafo:
                print(f"\n--- Recorridos desde {inicio} ---")
                busq_anch(grafo, inicio)
                print()
                busq_prof(grafo, inicio)
            else:

                print(f"Error: El centro '{inicio}' no existe.")
        elif opcion == "2":
            inicio = input("\nIngrese el centro desde donde iniciar: ").strip().upper()
            objetivo = input("Ingrese el centro que desea buscar: ").strip().upper()
            if inicio in grafo:
                busq_anch(grafo, inicio, objetivo)
            else:
                print(f"Error: El centro de inicio '{inicio}' no existe.")
        elif opcion == "3":
            print("\nGracias por utilizar el sistema de red de distribución. ¡Hasta luego!")
            break
        else:
            print("\nOpción no válida. Por favor, intente de nuevo.")
main()
```
