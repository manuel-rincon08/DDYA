# Merge Sort
En el merge sort divide el conjunto en 2 partes iguales hasta que cada subconjunto tenga 1 solo elemento. Luego la función merge recorre todos los subconjuntos para unir los elementos de manera ordenada
Teorema Maestro
T[n]= 2T(n/2)+Θ(n)

 Recurrencia:

 T[n]= Θ(n^log2(2)*log^(0+1)(n))

 T[n]= Θ(nlog(n))

Mejor caso = 1

Peor caso = nlog(n)

# Búsqueda binaria
En la búsqueda binaria funciona en una lista ordenada, donde en cada paso se divide en 2 partes iguales y se descarta la parte donde no esta el elemento que queremos encontrar
Teorema maestro
T[n]= T(n/2)+Θ(n)

Recurrencia:

T[n]= Θ(n^log2(1)*log^(0+1)(n))

T[n]= Θ(log(n))

Mejor caso =1

Peor caso =logn

```python
def compr(a):
    try:
        val=int(a)
        if val <= 0:
            raise ValueError
        print("Dato valido")
        return 1, val
    except ValueError:
        print("Dato invalido, por favor asegurese de ingresar solo: \n- Numeros enteros \n- Numeros positivos \n- No dejar el campo vacio  \n- No ingresar letras ni simbolos \n- Introducir un solo numero \n- No usar espacios, comas o cualquier otro caracter")
        return 0, 0

def merge_sort(cod):
    if len(cod) > 1:
        r = len(cod)//2
        L = cod[:r]
        M = cod[r:]

        merge_sort(L)
        merge_sort(M)
        i=0
        j=0
        k=0
        while i<len(L) or j<len(M):
            if i<len(L) and (j==len(M) or L[i]<M[j]):
                cod[k]=L[i]
                i+=1
            else:
                cod[k]=M[j]
                j+=1
            k+=1
    return cod

def busq_bin(cod,x):
    lo = 0
    hi = len(cod) - 1
    while lo <= hi:
        mid = (lo + hi) // 2
        if cod[mid] == x:
            print("Su numero se encuentra en la lista, en la posicion",(mid+1))
            b=0
            while b==0:
                c=input("Desea buscar otro numero? \nSi- Presione 1\nNo- Presione 2\n")
                b,c=compr(c)
            if c==1:
                return 0
            else:
                return 1
        elif cod[mid] > x:
            hi = mid - 1
        else:
            lo = mid + 1
    print("Su numero no se encuentra en la lista")
    return 0


def main():
    n=0
    b=0
    print("Bienvenido al programa")
    while b==0:
        n=input("Digite la cantida de productos que quiere ingresar:")
        b,n= compr(n)

    cod=[0 for c in range (n)]
    c=0
    while c<n:
        cod[c]=input("Digite el codigo del producto #"+str(c+1)+":")
        b,val=compr(cod[c])
        if b==1:
            cod[c]=val
            c+=1
    print("Organizando codigos...")
    print(merge_sort(cod))
    b=0
    c=0
    while b==0 or c==0:
        x=input("Digite el codigo del producto que quiere buscar:")
        b,x=compr(x)
        if b==1:
            c=busq_bin(cod,x)
    print("Gracias por usar el programa, hasta luego")
main()5
```
