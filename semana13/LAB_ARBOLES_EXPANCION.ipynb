## 1) Descripción del problema **[1 pts]**

Un equipo de electrónica está diseñando un prototipo sobre una protoboard. Sobre ella se han ubicado diferentes componentes que deben quedar conectados entre sí.

El problema es que se desea utilizar la menor cantidad posible de cable.

Para resolverlo, cada componente será representado como un vértice, cada posible conexión mediante cable será una arista y la longitud del cable será el peso de la arista.

El objetivo será construir un Árbol de Expansión Mínima (MST) que conecte todos los componentes utilizando la menor longitud total de cable.

Para este laboratorio no se evaluará el funcionamiento electrónico del circuito. La protoboard se utilizará como representación espacial del problema.

```python
import matplotlib.pyplot as plt

def coord(posic):
   fila = posic[0]
   columna = posic[1]
   fila_num = ord(fila.upper()) - 64
   return fila_num, columna

def dist(pos1, pos2):
   f1, c1 = coord(pos1)
   f2, c2 = coord(pos2)
   return abs(c1 - c2) + abs(f1 - f2)

def conex(comp):
   nom = list(comp.keys())
   conexiones = []
   for i in range(len(nom)):
       for j in range(i + 1, len(nom)):
           comp1 = nom[i]
           comp2 = nom[j]
           peso = dist(comp[comp1], comp[comp2])
           conexiones.append((peso, comp1, comp2))
   return conexiones

def padr(padre, i):
   if padre[i] == i:
       return i
   return padr(padre, padre[i])

def unir_nodos(padre, x, y):
   raiz_x = padr(padre, x)
   raiz_y = padr(padre, y)
   padre[raiz_x] = raiz_y

def kruskal(conexiones, nodos):
   conexiones.sort()
   padre = {nodo: nodo for nodo in nodos}
   mst = []
   longitud_total = 0
   for peso, u, v in conexiones:
       raiz_u = padr(padre, u)
       raiz_v = padr(padre, v)
       if raiz_u != raiz_v:
           mst.append((peso, u, v))
           longitud_total += peso
           unir_nodos(padre, raiz_u, raiz_v)
           if len(mst) == len(nodos) - 1:
               break
   return mst, longitud_total
def proto1(titulo):
   plt.figure(figsize=(10, 5))
   plt.yticks(range(1, 11), ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J'])
   plt.xticks(range(1, 31))
   plt.gca().invert_yaxis()
   plt.grid(True, linestyle='--', alpha=0.6)
   plt.title(titulo)
   plt.xlabel("Columnas")
   plt.ylabel("Filas")
   plt.show()

def proto(componentes, conexiones_mst, titulo):
    plt.figure(figsize=(10, 5))
    for nombre, pos in componentes.items():
        f, c = coord(pos)
        plt.scatter(c, f, s=150, zorder=5, label=nombre)
        plt.text(c, f - 0.3, nombre, fontsize=12, ha='center', weight='bold')
    for peso, u, v in conexiones_mst:
        f1, c1 = coord(componentes[u])
        f2, c2 = coord(componentes[v])
        plt.plot([c1, c2], [f1, f1], 'k-', zorder=2)
        plt.plot([c2, c2], [f1, f2], 'k-', zorder=2)
        plt.text(c2, f1 - 0.3, str(peso), color='red', fontsize=10, weight='bold')
    plt.yticks(range(1, 11), ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J'])
    plt.xticks(range(1, 31))
    plt.gca().invert_yaxis()
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.title(titulo)
    plt.xlabel("Columnas")
    plt.ylabel("Filas")
    plt.show()

def impr(mst, longitud_total):
   print("=========== RESULTADO ===========")
   for i, (peso, u, v) in enumerate(mst, 1):
       print(f"Cable {i}: {u} - {v} = {peso}")
   print(f"Longitud total de cable: {longitud_total} unidades")
   print("=================================\n")

def main():
   comp_or = {'F': ('A', 3), 'L': ('C', 8), 'S': ('B', 16),'P': ('D', 24), 'B': ('H', 6), 'R': ('I', 15), 'M': ('G', 25)}
   nodos = list(comp_or.keys())
   tod_conex = conex(comp_or)
   print("Todas las posibles conexiones")
   for peso, u, v in tod_conex:
       print(f"{u} -------- {peso} -------- {v}")
   proto1("protoboard")
   print("\n")
   print("MST Original")
   mst_ori, longitud_original = kruskal(tod_conex, nodos)
   impr(mst_ori, longitud_original)
   proto(comp_or, mst_ori, "Protoboard - Distribución Original (MST)")
   nuev_comp = comp_or.copy()
   nuev_comp['B'] = ('J', 20)
   nuev_conex = conex(nuev_comp)
   nuev_mst, longitud_nueva = kruskal(nuev_conex.copy(), nodos)
   print("--- MST Después del Cambio (Punto 6) ---")
   print(f"MST original Longitud total: {longitud_original}")
   print(f"MST después del cambio Longitud total: {longitud_nueva}")
   if longitud_nueva > longitud_original:
       print("La cantidad total de cable necesaria AUMENTÓ.")
   elif longitud_nueva < longitud_original:
       print("La cantidad total de cable necesaria DISMINUYÓ.")
   else:
       print("La cantidad total de cable necesaria SE MANTUVO IGUAL.")
   proto(nuev_comp, nuev_mst, "Protoboard - Distribución Actualizada (MST)")
main()
```
