## 6) Código funcional **[15 pts]**
Incluye tu implementación completa.

No pegues código incompleto; debe ser ejecutable.


```python
memoria={}
def plantas_producidas(n,memoria):
    if n < 0 :
        return 'Dato inválido,escriba un número positivo'
    if n == 0:
        return 0
    if n==1:
        return 1
    if n in memoria:
        return memoria[n]
    resultado = plantas_producidas(n-1,memoria)+ plantas_producidas(n-2,memoria)
    return resultado
def main():
    resultado = plantas_producidas(n,memoria)
    print(resultado)
```
